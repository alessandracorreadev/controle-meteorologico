```python
endpoint_sheet = "https://docs.google.com/spreadsheets/d/10QtSot_yhoJRZTNOKgQHBmFODRiVlU-ko7J7p5hBWfY/export?format=csv"

endpoint_api = "https://api.open-meteo.com/v1/forecast"

print("----------Dados Coletados----------")

dados_planilha = pd.read_csv(endpoint_sheet)

display(dados_planilha)
```

```python
for indice, linha in dados_planilha.iterrows():
  nome_local = linha['Nome do local']
  lat = linha['Latitude']
  lon = linha['Longitude']

  parametros = {
      'latitude': lat,
      'longitude': lon,
      'current': 'temperature_2m'
  }

  resposta = requests.get(endpoint_api, params=parametros).json()

  display(resposta)
```
